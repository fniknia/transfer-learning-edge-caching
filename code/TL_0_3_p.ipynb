```python
class SumTree(object):
    data_pointer = 0

    # Here we initialize the tree with all nodes = 0, and initialize the data with all values = 0
    def __init__(self, capacity, no_permanent_data, n_step, gamma):
        # Number of leaf nodes (final nodes) that contains experiences
        self.capacity = capacity
        self.no_permanent_data = no_permanent_data
        self.n_step = n_step

        # Generate the tree with all nodes values = 0
        # To understand this calculation (2 * capacity - 1) look at the schema below
        # Remember we are in a binary node (each node has max 2 children) so 2x size of leaf (capacity) - 1 (root node)
        # Parent nodes = capacity - 1
        # Leaf nodes = capacity
        self.tree = np.zeros(2 * capacity - 1)

        # Contains the experiences (so the size of data is capacity)
        self.data = np.zeros(capacity, dtype=object)
        self.full = 0

        self.gamma = gamma

    def add(self, priority, data, TrialNo, copying):
        # Look at what index we want to put the experience
        tree_index = self.data_pointer + self.capacity - 1

        #n-step
        n_step_reward = 0
        time = 0

        if(TrialNo > self.n_step) and (copying == 0):
            for i in range(self.n_step - 1):
                n_step_index = (self.data_pointer - self.n_step + i) % self.capacity
                #print("n_step_index:", n_step_index)
                #print("self.data:", self.data[n_step_index])
                time += self.data[n_step_index][4]
                n_step_reward += (self.gamma ** time) * self.data[n_step_index][2]

            n_step_state = data[0]
            n_step_transition_time = time + self.data[(self.data_pointer - 1) % self.capacity][4]

            idx = (self.data_pointer - self.n_step) % self.capacity

            self.data[idx] = self.data[idx][0], self.data[idx][1], self.data[idx][2], self.data[idx][3], self.data[idx][4], \
            self.data[idx][5], self.data[idx][6], n_step_reward, n_step_state, n_step_transition_time, 0

            #experience = state0, action1, reward2, nextstate3, transition_time4, value5, log_prob6, n_step_reward7, n_step_state8, n_step_transition_time9, IsDemo0

        # Update data frame
        self.data[self.data_pointer] = data

        # Update the leaf
        self.update (tree_index, priority)

        # Add 1 to data_pointer
        self.data_pointer += 1

        if self.data_pointer >= self.capacity:  # If we're above the capacity, we go back to first index (we overwrite)
            self.data_pointer = self.no_permanent_data  # make sure demo data permanent
            self.full = 1

    def update(self, tree_index, priority):
        #print("priority:", priority)
        # Change = new priority score - former priority score
        change = priority - self.tree[tree_index]
        self.tree[tree_index] = priority

        # then propagate the change through tree
        # this method is faster than the recursive loop
        while tree_index != 0:
            tree_index = (tree_index - 1) // 2
            self.tree[tree_index] += change

    def get_leaf(self, v):
        parent_index = 0

        while True:
            left_child_index = 2 * parent_index + 1
            right_child_index = left_child_index + 1

            # If we reach bottom, end the search
            if left_child_index >= len(self.tree):
                leaf_index = parent_index
                break
            else: # downward search, always search for a higher priority node
                if v <= self.tree[left_child_index]:
                    parent_index = left_child_index
                else:
                    v -= self.tree[left_child_index]
                    parent_index = right_child_index

        data_index = leaf_index - self.capacity + 1

        return leaf_index, self.tree[leaf_index], self.data[data_index], data_index

    @property
    def total_priority(self):
        return self.tree[0] # Returns the root node
```

```python
class Memory(object):  # stored as ( state, action, reward, next_state ) in SumTree
    epsilon = 0.001  # small amount to avoid zero priority
    demo_epsilon = 1.0  # 1.0  # extra
    alpha = 0.4  # [0~1] convert the importance of TD error to priority
    beta = 0.6  # importance-sampling, from initial value increasing to 1
    beta_increment_per_sampling = 0.001
    abs_err_upper = 1.  # clipped abs error

    def __init__(self, capacity, gamma):
        # Making the tree
        self.n_step = 10
        self.tree_target = SumTree(capacity, 0, self.n_step, gamma)
        self.capacity = capacity

    def copy_demo(self, TrialNo):
        #for i in range(self.capacity):
            with h5py.File('states.h5', "r") as f:
                states = f['data1'][()]

            with h5py.File('next_states.h5', "r") as f:
                next_states = f['data1'][()]

            with h5py.File('actions.h5', "r") as f:
                actions = f['data1'][()]

            with h5py.File('rewards.h5', "r") as f:
                rewards = f['data1'][()]

            with h5py.File('transition_times.h5', "r") as f:
                transition_times = f['data1'][()]

            with h5py.File('n_step_rewards.h5', "r") as f:
                n_step_rewards = f['data1'][()]

            with h5py.File('n_step_states.h5', "r") as f:
                n_step_states = f['data1'][()]

            with h5py.File('n_step_transition_times.h5', "r") as f:
                n_step_transition_times = f['data1'][()]

            with h5py.File('priprities.h5', "r") as f:
                priprities = f['data1'][()]

            with h5py.File('values.h5', "r") as f:
                values = f['data1'][()]

            with h5py.File('log_probs.h5', "r") as f:
                log_probs = f['data1'][()]

            print("reading")

            for i in range(len(rewards)):
                #experience = state0, action1, reward2, nextstate3, transition_time4, value5, log_prob6, n_step_reward7, n_step_state8, n_step_transition_time9, IsDemo0
                Demo_transition = states[i], actions[i], rewards[i], next_states[i], transition_times[i], values[i], \
                log_probs[i], n_step_rewards[i], n_step_states[i], n_step_transition_times[i], 1
                # 1 indicates that this is a demo data, 0 indicates the number of the times that the transition is sampled

                self.tree_target.add(priprities[i], Demo_transition, TrialNo, 1)

            #print("Demo_transition:", np.shape(Demo_transition))
            print("done with copying")

    def store(self, experience, trial_no):
        # Find the max priority
        max_priority = np.max(self.tree_target.tree[-self.tree_target.capacity:])

        # If the max priority = 0 we can't put priority = 0 since this experience will never have a chance to be selected
        # So we use a minimum priority
        if max_priority == 0:
            max_priority = self.abs_err_upper

        #print("store agent shape:", np.shape(experience))
        #print("experience:", experience[9])

        #print("exp len:", len(experience))
        self.tree_target.add(max_priority, experience, trial_no, 0)   # set the max priority for new priority

    def sample(self, n, trial_no):
        # Create a minibatch array that will contains the minibatch
        minibatch = []
        b_idx = np.empty((n,), dtype=np.int32)
        data_idx = np.empty((n,), dtype=np.int32)
        self.beta = np.min([1., self.beta + self.beta_increment_per_sampling])
        ISWeights = []

        priority_segment = self.tree_target.total_priority / n       # priority segment

        min_prob = np.min(self.tree_target.tree[-self.tree_target.capacity:]) / self.tree_target.total_priority
        assert min_prob > 0

        for i in range(n):
            # A value is uniformly sample from each range
            a, b = priority_segment * i, priority_segment * (i + 1)
            value = np.random.uniform(a, b)

            # Experience that correspond to each value is retrieved
            index, priority, data, data_index = self.tree_target.get_leaf(value)
            b_idx[i] = index
            data_idx[i] = data_index
            minibatch.append([data[0], data[1], data[2], data[3], data[4], data[5], data[6], data[7], data[8], data[9], data[10]])

            #samp_count = data[9] + 1
            self.tree_target.data[data_index] = data[0], data[1], data[2], data[3], data[4], data[5], data[6], data[7], data[8], data[9], data[10]

            prob = priority / self.tree_target.total_priority
            ISWeights.append(np.power(prob/min_prob, -self.beta))

        return b_idx, minibatch, ISWeights, data_idx

    def batch_update(self, tree_idx, abs_errors, trial_no):
        abs_errors += self.epsilon  # convert to abs and avoid 0
        abs_errors[:self.tree_target.no_permanent_data] += self.demo_epsilon
        clipped_errors = np.minimum(abs_errors, self.abs_err_upper)
        ps = np.power(clipped_errors, self.alpha)

        for ti, p in zip(tree_idx, ps):
            self.tree_target.update(ti, p)
```

```python
# Actor network with He initialization\
def build_actor(learning_rate, n_actions, input_dims, fc1_dims, fc2_dims):
    actor = Sequential([
        Input(shape=(input_dims,)),
        Dense(128, activation='relu', kernel_initializer=HeNormal()),
        Dense(128, activation='relu', kernel_initializer=HeNormal()),
        Dense(n_actions, kernel_initializer=HeNormal())
    ])

    return actor

# Critic network with He initialization
def build_critic(learning_rate, n_actions, input_dims, fc1_dims, fc2_dims):
    critic = Sequential([
        Input(shape=(input_dims,)),
        Dense(input_dims, activation='relu', kernel_initializer=HeNormal()),
        Dense(128, activation='relu', kernel_initializer=HeNormal()),
        Dense(1, kernel_initializer=HeNormal())
    ])

    return critic
```

```python
class Agent(object):
    def __init__(self, n_actions, mem_size, input_dims):
        self.learning_rate = 0.0005
        self.gamma = 0.99
        self.n_actions = n_actions
        self.epsilon = 1
        self.epsilon_dec = 0.999
        self.epsilon_min = 0.01
        self.batch_size = 64
        self.input_dims = input_dims
        self.mem_size = mem_size
        self.update_target = 100
        self.λ_dq = 1
        self.λ_n_dq = 1
        self.λ_jeq = 1
        self.λ_l2 = 0.00001
        self.n_step = 10
        self.clip_ratio = 0.2  # Clipping parameter for PPO
        self.lr_actor = 0.0001  # Learning rate for actor
        self.lr_critic = 0.0005  # Learning rate for critic
        self.entropy_coef = 0.01

        self.MEMORY = Memory(self.mem_size, self.gamma)

        self.ISWeights = np.zeros(self.batch_size, dtype = float)
        self.tree_idx = np.zeros(self.batch_size, dtype = float)
        self.data_idx = np.zeros(self.batch_size, dtype = float)
        self.isdemo = np.zeros(self.batch_size, dtype = float)

        self.actor = build_actor(self.learning_rate, self.n_actions, input_dims, 16, 16)
        self.critic = build_critic(self.learning_rate, self.n_actions, input_dims, 16, 16)

        self.values_next = np.zeros((self.batch_size))
        self.weights = np.zeros((self.batch_size), dtype = np.float32)
        self.actions = np.zeros((self.batch_size))
        self.states = np.zeros((self.batch_size,  self.input_dims))
        self.n_step_states = np.zeros((self.batch_size, self.input_dims))
        self.returns = np.zeros((self.batch_size))
        self.n_step_returns = np.zeros((self.batch_size))

        self.weights_online = 0
        self.weights_target = 0

        self.MEMORY.copy_demo(1)

        self.actor.compile(optimizer='adam')
        self.critic.compile(optimizer='adam', loss=self.custom_loss)

        self.actor_optimizer = tf.keras.optimizers.Adam(learning_rate = self.lr_actor)
        self.critic_optimizer = tf.keras.optimizers.Adam(learning_rate = self.lr_critic)

    def get_action_and_value(self, state):
        #state = state.reshape(1, -1)
        logits = self.actor.predict(state, verbose = 0)
        action_prob = tf.nn.softmax(logits)

        action = tf.random.categorical(logits, 1)[0, 0]
        log_prob = tf.nn.log_softmax(logits)[0, action]

        value = self.critic.predict(state, verbose = 0)[0, 0]

        return action, value, log_prob

    def store(self, state, action, reward, transition_time, nextstate, i, No_source_requests, value, log_prob, trial_no):
        if((agent.trial  - 1) >= L):
            return

        experience = state, action, reward, nextstate, transition_time, value, log_prob, 0, np.zeros((2 + FileTypes * 8)), 0, 0
        self.MEMORY.store(experience, trial_no)

    def learn(self, trial_no, avg_R):
        if(self.MEMORY.tree_target.full == 1):
            self.tree_idx, minibatch, self.ISWeights, self.data_idx = self.MEMORY.sample(self.batch_size, trial_no)

            next_states = np.zeros((self.batch_size, self.input_dims))
            rewards = np.zeros((self.batch_size))
            n_step_rewards = np.zeros((self.batch_size))
            transition_times = np.zeros((self.batch_size))
            n_step_transition_times = np.zeros((self.batch_size))
            values = np.zeros((self.batch_size))
            absolute_errors = np.zeros((self.batch_size))
            log_probs = []

            for i in range(self.batch_size):
                self.states[i] = minibatch[i][0]
                #print("self.states[i]:", np.shape(self.states[i]))
                self.actions[i] = minibatch[i][1]
                rewards[i] = minibatch[i][2]
                next_states[i] = minibatch[i][3]
                transition_times[i] = minibatch[i][4]
                values[i] = minibatch[i][5]
                log_probs.append(minibatch[i][6])
                n_step_rewards[i] = minibatch[i][7]
                self.n_step_states[i] = minibatch[i][8]
                n_step_transition_times[i] = minibatch[i][9]
                self.isdemo[i] = minibatch[i][10]

            self.values_next = self.critic.predict(next_states, verbose = 0)
            n_step_values_next = self.critic.predict(self.n_step_states, verbose = 0)

            for i in range(self.batch_size):
                self.returns[i] = rewards[i] + (self.gamma ** transition_times[i]) * self.values_next[i, 0]
                self.n_step_returns[i] = n_step_rewards[i] + (self.gamma ** n_step_transition_times[i]) * n_step_values_next[i, 0]


            #print("self.returns:", self.returns)
            #print("self.n_step_returns:", self.n_step_returns)

            advantages = self.returns - values

            #training
            with tf.GradientTape() as tape:
                logits = self.actor(self.states)
                #print("shape:", np.shape(states))
                action_dist = tf.nn.softmax(logits)
                action_dist = tf.clip_by_value(action_dist, 1e-10, 1.0) # Ensure values are not exactly zero
                new_log_probs = tf.reduce_sum(tf.math.log(action_dist) * tf.one_hot(self.actions, self.n_actions), axis=1)
                ratio = tf.exp(new_log_probs - log_probs)
                clipped_ratio = tf.clip_by_value(ratio, 1 - self.clip_ratio, 1 + self.clip_ratio)
                policy_loss = -tf.reduce_mean(tf.minimum(ratio * advantages, clipped_ratio * advantages))

            actor_grads = tape.gradient(policy_loss, self.actor.trainable_variables)
            self.actor_optimizer.apply_gradients(zip(actor_grads, self.actor.trainable_variables))
            #####
            train = self.critic_optimizer.minimize(self.custom_loss, var_list=self.critic.trainable_variables)
            #self.critic_optimizer.apply_gradients(zip(critic_grads, self.critic.trainable_variables))

            absolute_errors = np.abs(self.returns - values) + np.abs(avg_R - rewards)
            #absolute_errors = np.abs(self.returns - values)
            #print("absolute_errors:", absolute_errors)


            # Update priority
            self.MEMORY.batch_update(self.tree_idx, absolute_errors, trial_no)

            self.epsilon = max(self.epsilon_min, self.epsilon * self.epsilon_dec)

    def loss_l(self, ae, a):
        return 0.0 if ae == a else 0.8

    def loss_jeq(self):
        jeq = 0.0
        for i in range(self.batch_size):
            ae = self.actions[i]
            max_value = float("-inf")
            for a in range(self.n_actions):
                max_value = np.maximum(self.values_next[i] + self.loss_l(ae, a), max_value)

            jeq += self.isdemo[i] * (max_value - self.values_next[i])
        return jeq

    def custom_loss(self):
        with tf.GradientTape() as tape:
            values = self.critic(self.states)
            n_step_values = self.critic(self.n_step_states)
            #print("values:", values)
            #value_loss = self.ISWeights * tf.reduce_mean((returns - values) ** 2)

            l_dq = tf.reduce_mean(tf.math.squared_difference(tf.cast(self.returns, tf.float32), tf.cast(values, tf.float32)))
            l_n_dq = tf.reduce_mean(tf.math.squared_difference(tf.cast(self.n_step_returns, tf.float32), tf.cast(n_step_values, tf.float32)))
            l_jeq = tf.cast(self.loss_jeq(), tf.float32)
            l_l2 = tf.reduce_sum([tf.reduce_mean(reg_l) for reg_l in tf.compat.v1.get_collection(tf.compat.v1.GraphKeys.REGULARIZATION_LOSSES)])
            value_loss = tf.cast(self.ISWeights, tf.float32) * tf.reduce_sum(self.λ_dq * l_dq + self.λ_n_dq * l_n_dq + self.λ_jeq * l_jeq + self.λ_l2 * l_l2)

        #gradients = tape.gradient(value_loss, self.critic)
        #print("gradients:", gradients)
        return value_loss
```

```python
def Update_Content(Cache, t, Cachesize):
    RExpire = 0
    exit = 0
    ExpiredFile_hitcount = 0
    for i in range(Cachesize):
        if(Cache[i][0] != 0):
            if(Cache[i][2] + Cache[i][3] <= t):
                Cache[i][:] = 0

    return Cache, RExpire, exit, ExpiredFile_hitcount
```

```python
def Update_Freshness(Cache, Current, CacheSize):
    for i in range(CacheSize):
        if(Cache[i][0] != 0):
            Cache[i][4] = (Current - Cache[i][2]) / Cache[i][3]
            Cache[i][5] = (Cache[i][2] + Cache[i][3] - Current) / Cache[i][3]
            Cache[i][6] = (Cache[i][2] + Cache[i][3] - Current)
    return Cache
```

```python
def Update_Utility(Cache, CacheSize):
    max_utility = 1.5
    fresh_importance = 0.1
    fresh = 0
    min_utility = max_utility * (1 - fresh_importance)
    curvature = (max_utility - min_utility) / (e - 1)

    for i in range(CacheSize):
        if(Cache[i][0] != 0):
            Cache[i][8] = (-e ** (Cache[i][4] + log(curvature)) + max_utility + curvature) * Cache[i][7]
    return Cache
```

```python
def sum_hitcount_short(Hitcount_short, FileTypes):
    s = np.zeros((FileTypes), dtype = np.int32)

    for i in range(FileTypes):
        s[i] = np.sum(Hitcount_short[:, i])

    return s
```

```python
def DoAction(Action, RequestedFile, Cache1, Cache2, Cachesize, t, LifeT, importance, size, CacheCapacity):
    flag = 0

    max_utility = 1.5
    fresh_importance = 0.1
    fresh = 0
    min_utility = max_utility * (1 - fresh_importance)
    curvature = (max_utility - min_utility) / (e - 1)
    file_utility = (-e ** (fresh + log(curvature)) + max_utility + curvature) * importance

    #If the file is already cached.
    for j in range(Cachesize):
        if (Cache1[j, 0] == RequestedFile or Cache2[j, 0] == RequestedFile):
            if(Cache1[j, 0] == RequestedFile):
                Cache1[j, 1] += 1
            else:
                Cache2[j, 1] += 1
            flag = 1
            break

    if(Action == 1 and flag == 0):
        if(size <= CacheCapacity - np.sum(Cache1[:, 9])):
            for i in range(Cachesize):
                if(Cache1[i, 0] == 0):
                    Cache1[i, 0] = RequestedFile
                    Cache1[i, 1] = 0 #hit
                    Cache1[i, 2] = t #generation time
                    Cache1[i, 3] = LifeT
                    Cache1[i, 4] = 0 #normalized age
                    Cache1[i, 5] = 1 #normalized remaining lifetime
                    Cache1[i, 6] = LifeT #remaining lifetime
                    Cache1[i, 7] = importance
                    Cache1[i, 8] = file_utility #utility
                    Cache1[i, 9] = size #size
                    break
        else:

            cache_temp = np.vstack((np.transpose(Cache1[:, 8]), np.transpose(Cache1[:, 0])))
            tr_cache_temp = np.transpose(cache_temp)
            cache_temp_sort = tr_cache_temp[tr_cache_temp[:, 0].argsort()]

            j = 0
            while(size > CacheCapacity - np.sum(Cache1[:, 9])):
                for i in range(Cachesize):
                    if(Cache1[i, 0] == cache_temp_sort[j, 1]):
                        Cache1[i][:] = 0

                j += 1

            for i in range(Cachesize):
                if(Cache1[i, 0] == 0):
                    Cache1[i][0] = RequestedFile
                    Cache1[i][1] = 0
                    Cache1[i][2] = t
                    Cache1[i][3] = LifeT
                    Cache1[i][4] = 0
                    Cache1[i][5] = 1
                    Cache1[i][6] = LifeT
                    Cache1[i][7] = importance
                    Cache1[i][8] = file_utility
                    Cache1[i][9] = size
                    break

    elif(Action == 2 and flag == 0):
        if(size <= CacheCapacity - np.sum(Cache2[:, 9])):
            for i in range(Cachesize):
                if(Cache2[i, 0] == 0):
                    Cache2[i, 0] = RequestedFile
                    Cache2[i, 1] = 0 #hit
                    Cache2[i, 2] = t #generation time
                    Cache2[i, 3] = LifeT
                    Cache2[i, 4] = 0 #normalized age
                    Cache2[i, 5] = 1 #normalized remaining lifetime
                    Cache2[i, 6] = LifeT #remaining lifetime
                    Cache2[i, 7] = importance
                    Cache2[i, 8] = file_utility #utility
                    Cache2[i, 9] = size #size
                    break
        else:

            cache_temp = np.vstack((np.transpose(Cache2[:, 8]), np.transpose(Cache2[:, 0])))
            tr_cache_temp = np.transpose(cache_temp)
            cache_temp_sort = tr_cache_temp[tr_cache_temp[:, 0].argsort()]

            j = 0
            while(size > CacheCapacity - np.sum(Cache2[:, 9])):
                for i in range(Cachesize):
                    if(Cache2[i, 0] == cache_temp_sort[j, 1]):
                        Cache2[i][:] = 0

                j += 1

            for i in range(Cachesize):
                if(Cache2[i, 0] == 0):
                    Cache2[i][0] = RequestedFile
                    Cache2[i][1] = 0
                    Cache2[i][2] = t
                    Cache2[i][3] = LifeT
                    Cache2[i][4] = 0
                    Cache2[i][5] = 1
                    Cache2[i][6] = LifeT
                    Cache2[i][7] = importance
                    Cache2[i][8] = file_utility
                    Cache2[i][9] = size
                    break

    return Cache1, Cache2, flag, History_short
```

```python
def StateCache(Cache, Cachesize, FileTypes):
    History_short_100 = sum_hitcount_short(History_short, FileTypes)

    presents = np.zeros((FileTypes))
    utilities = np.zeros((FileTypes))

    for i in range(Cachesize):
        if(Cache[i, 0] != 0):
            presents[i] = 1
            utilities[i] = Cache[i, 8]

    return presents, utilities
```

```python
def Reward(Ut, Cache1, Cache2, History_short_100, FileTypes, CacheCapacity, TrialNo):

        Ut = 0
        utility_current = np.sum(Cache1[:, 8])
        r = 0
        for i in range(Cachesize):
            if(Cache1[i, 0] != 0):
                r += (max(1, History_short_100[int(Cache1[i, 0]) - 1])) * Cache1[i, 8]

                if(Cache1[i, 0] == RequestedFile):

                    #if(TrialNo < L):
                        Ut += Cache1[i, 8]

        r1 = r - ((CacheCapacity - np.sum(Cache1[:, 9])) / CacheCapacity) * 100


        utility_current = np.sum(Cache2[:, 8])
        r = 0
        for i in range(Cachesize):
            if(Cache2[i, 0] != 0):
                r += (max(1, History_short_100[int(Cache2[i, 0]) - 1])) * Cache2[i, 8]

                if(Cache2[i, 0] == RequestedFile):

                    #if(TrialNo < L):
                        Ut += 0.5 * Cache2[i, 8]

        r2 = r - ((CacheCapacity - np.sum(Cache2[:, 9])) / CacheCapacity) * 100

        reward = r1 + 0.5 * r2

        return reward, Ut
```
