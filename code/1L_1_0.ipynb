```python
class ReplayBuffer(object):
    def __init__(self, max_size, input_dims, n_actions, gamma, batch_size):
        self.n_step = 10
        self.gamma = gamma
        self.priority_epsilon = 0.001
        self.priority_epsilon_demo = 1
        self.gamma = gamma
        self.beta = 0.6
        self.beta_max = 1
        self.beta_inc = 0.001
        self.batch_size = batch_size
        self.input_dims = input_dims
        self.alpha = 0.4

        self.mem_size = max_size
        self.mem_cntr = 0
        self.state_memory = np.zeros((self.mem_size, self.input_dims))
        self.nextstate_memory = np.zeros((self.mem_size, self.input_dims))
        self.action_memory = np.zeros((self.mem_size), dtype = np.int32)
        self.reward_memory = np.zeros(self.mem_size)
        self.transition_time_memory = np.zeros(self.mem_size, dtype = np.float32)
        self.value = np.zeros(self.mem_size)
        self.log_prob = np.zeros(self.mem_size)

    def store_transition(self, state, action, reward, transition_time, nextstate, value, log_prob):
        index = self.mem_cntr % self.mem_size
        self.state_memory[index]= state
        self.nextstate_memory[index] = nextstate
        self.action_memory[index] = action
        self.reward_memory[index] = reward
        self.transition_time_memory[index] = transition_time
        self.value[index] = value
        self.log_prob[index] = log_prob

        self.mem_cntr += 1

    def random_sample_source(self, batch_size):
        max_mem = min((self.mem_cntr), self.mem_size)
        batch = np.random.randint(low=0, high=max_mem, size=batch_size)

        states = self.state_memory[batch, :]
        nextstates = self.nextstate_memory[batch, :]
        actions = self.action_memory[batch]
        rewards = self.reward_memory[batch]
        transition_times = self.transition_time_memory[batch]
        value = self.value[batch]
        log_prob = self.log_prob[batch]

        return states, actions, rewards, transition_times, nextstates, batch, log_prob, value
```

```python
# Actor network with He initialization\
def build_actor(learning_rate, n_actions, input_dims, fc1_dims, fc2_dims):
    actor = Sequential([
        Input(shape=(input_dims,)),
        Dense(128, activation='relu', kernel_initializer=HeNormal()),
        Dense(128, activation='relu', kernel_initializer=HeNormal()),
        Dense(n_actions, kernel_initializer=HeNormal())
    ])

    return actor

# Critic network with He initialization
def build_critic(learning_rate, n_actions, input_dims, fc1_dims, fc2_dims):
    critic = Sequential([
        Input(shape=(input_dims,)),
        Dense(128, activation='relu', kernel_initializer=HeNormal()),
        Dense(128, activation='relu', kernel_initializer=HeNormal()),
        Dense(1, kernel_initializer=HeNormal())
    ])

    return critic
```

```python
def Update_Content(Cache, t, Cachesize):
    RExpire = 0
    exit = 0
    ExpiredFile_hitcount = 0
    for i in range(Cachesize):
        if(Cache[i][0] != 0):
            if(Cache[i][2] + Cache[i][3] <= t):
                Cache[i][:] = 0

    return Cache, RExpire, exit, ExpiredFile_hitcount
```

```python
def Update_Freshness(Cache, Current, CacheSize):
    for i in range(CacheSize):
        if(Cache[i][0] != 0):
            Cache[i][4] = (Current - Cache[i][2]) / Cache[i][3]
            Cache[i][5] = (Cache[i][2] + Cache[i][3] - Current) / Cache[i][3]
            Cache[i][6] = (Cache[i][2] + Cache[i][3] - Current)
    return Cache
```

```python
def Update_Utility(Cache, CacheSize):
    max_utility = 1.5
    fresh_importance = 0.1
    fresh = 0
    min_utility = max_utility * (1 - fresh_importance)
    curvature = (max_utility - min_utility) / (e - 1)

    for i in range(CacheSize):
        if(Cache[i][0] != 0):
            Cache[i][8] = (-e ** (Cache[i][4] + log(curvature)) + max_utility + curvature) * Cache[i][7]
    return Cache
```

```python
def sum_hitcount_short(Hitcount_short, FileTypes):
    s = np.zeros((FileTypes), dtype = np.int32)

    for i in range(FileTypes):
        s[i] = np.sum(Hitcount_short[:, i])

    return s
```
